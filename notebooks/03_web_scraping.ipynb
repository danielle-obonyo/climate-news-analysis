# Part 1: Web Scraping

## Objective

The objective of this project is to collect at least 50
news articles related to climate change and renewable energy.

## Selected Website

The Guardian

## Selected Tools

- Requests: Download RSS feeds and webpages.
- BeautifulSoup: Parse RSS and HTML documents.
- Trafilatura: Extract the main article text.
- Pandas: Store and export article information.

## Collection Method

RSS feeds will be used to discover relevant articles.
The article URLs will then be downloaded and their main
text extracted using Trafilatura.

```python
import requests
import pandas as pd
import time
import trafilatura

from bs4 import BeautifulSoup
from urllib.parse import urlparse

print("Web scraping libraries ready!")
```

```python
rss_feeds = {
    "Climate Change":
        "https://www.theguardian.com/environment/climate-crisis/rss",

    "Renewable Energy":
        "https://www.theguardian.com/environment/renewableenergy/rss",

    "Environment":
        "https://www.theguardian.com/environment/rss",

    "Science":
        "https://www.theguardian.com/science/rss",

    "World News":
        "https://www.theguardian.com/world/rss"
}


headers = {
    "User-Agent": "StudentResearchProject/1.0 (educational research)"
}

articles_found = []

for topic, feed_url in rss_feeds.items():

    try:
        response = requests.get(
            feed_url,
            headers=headers,
            timeout=30
        )

        response.raise_for_status()

        soup = BeautifulSoup(response.content, "xml")

        items = soup.find_all("item")

        print(f"\n{topic}: {len(items)} articles found")

        for item in items:

            title = item.find("title")
            link = item.find("link")
            pub_date = item.find("pubDate")

            if title and link:

                articles_found.append({
                    "topic": topic,
                    "title": title.get_text(strip=True),
                    "url": link.get_text(strip=True),
                    "publication_date": (
                        pub_date.get_text(strip=True)
                        if pub_date else None
                    )
                })

    except requests.RequestException as e:
        print(f"Error accessing {topic}: {e}")

articles_df = pd.DataFrame(articles_found)

articles_df = articles_df.drop_duplicates(
    subset=["url"]
).reset_index(drop=True)

print("\nUnique articles discovered:", len(articles_df))

display(articles_df.head(10))
```

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup

headers = {
    "User-Agent": "StudentResearchProject/1.0 (educational research)"
}

keywords = [
    "climate",
    "renewable energy",
    "solar energy",
    "wind power",
    "global warming",
    "greenhouse gas",
    "carbon emissions",
    "fossil fuel",
    "clean energy",
    "sustainability",
    "environment",
    "electric vehicle",
    "biodiversity",
    "deforestation",
    "pollution"
]

articles_found = []

for topic, feed_url in rss_feeds.items():

    try:
        response = requests.get(
            feed_url,
            headers=headers,
            timeout=30
        )

        response.raise_for_status()

        soup = BeautifulSoup(response.content, "xml")

        items = soup.find_all("item")

        count = 0

        for item in items:

            title_tag = item.find("title")
            link_tag = item.find("link")
            date_tag = item.find("pubDate")
            description_tag = item.find("description")

            if not title_tag or not link_tag:
                continue

            title = title_tag.get_text(strip=True)
            url = link_tag.get_text(strip=True)

            description = (
                description_tag.get_text(" ", strip=True)
                if description_tag else ""
            )

            searchable_text = (
                title + " " + description
            ).lower()

            if any(word in searchable_text for word in keywords):

                articles_found.append({
                    "topic": topic,
                    "title": title,
                    "url": url,
                    "description": description,
                    "publication_date": (
                        date_tag.get_text(strip=True)
                        if date_tag else None
                    )
                })

                count += 1

        print(f"{topic}: {count} relevant articles")

    except requests.RequestException as e:
        print(f"Could not access {topic}: {e}")

articles_df = pd.DataFrame(articles_found)

articles_df = articles_df.drop_duplicates(
    subset=["url"]
).reset_index(drop=True)

print("\nTotal unique relevant articles:", len(articles_df))
```

```python
print("Total articles:", len(articles_df))

print("\nArticles by feed:")
print(articles_df["topic"].value_counts())

display(articles_df.head(10))
```

```python
articles_df.to_csv(
    "../data/discovered_articles.csv",
    index=False
)

print("Article discovery dataset saved!")
```

```python
#create extraction code
import requests
import trafilatura
import pandas as pd
import time

from bs4 import BeautifulSoup
from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry

# Create a session with retry support
session = requests.Session()

retries = Retry(
    total=3,
    backoff_factor=2,
    status_forcelist=[429, 500, 502, 503, 504],
    allowed_methods=["GET"]
)

session.mount("https://", HTTPAdapter(max_retries=retries))

headers = {
    "User-Agent": "StudentResearchProject/1.0 (educational research)"
}

# Store extracted articles
scraped_articles = []

# Track failed articles
failed_articles = []

print("Starting article extraction...\n")

for index, row in articles_df.iterrows():

    url = row["url"]

    try:
        response = session.get(
            url,
            headers=headers,
            timeout=30
        )

        response.raise_for_status()

        # Extract the main article text
        article_text = trafilatura.extract(
            response.text,
            include_comments=False,
            include_tables=False,
            favor_precision=True
        )

        if article_text and len(article_text.split()) >= 100:

            scraped_articles.append({
                "topic": row["topic"],
                "title": row["title"],
                "url": url,
                "publication_date": row["publication_date"],
                "article_text": article_text,
                "word_count": len(article_text.split()),
                "extraction_status": "Success"
            })

            print(
                f"{index + 1}/{len(articles_df)} "
                f"SUCCESS: {row['title'][:55]}"
            )

        else:

            failed_articles.append({
                "url": url,
                "title": row["title"],
                "reason": "Insufficient article text"
            })

            print(
                f"{index + 1}/{len(articles_df)} "
                "SKIPPED: Insufficient article text"
            )

    except requests.RequestException as e:

        failed_articles.append({
            "url": url,
            "title": row["title"],
            "reason": str(e)
        })

        print(f"{index + 1}/{len(articles_df)} FAILED: {e}")

    except Exception as e:

        failed_articles.append({
            "url": url,
            "title": row["title"],
            "reason": str(e)
        })

        print(f"{index + 1}/{len(articles_df)} ERROR: {e}")

    # Be polite to the website
    time.sleep(1.5)

print("\nExtraction completed!")
```

```python
#check extraction results 
scraped_df = pd.DataFrame(scraped_articles)

failed_df = pd.DataFrame(failed_articles)

print("Extraction summary")
print("------------------")

print("Articles discovered:", len(articles_df))
print("Successfully extracted:", len(scraped_df))
print("Failed or skipped:", len(failed_df))

print(
    "\nSuccess rate:",
    round(len(scraped_df) / len(articles_df) * 100, 2),
    "%"
)

display(
    scraped_df[
        ["title", "topic", "word_count", "extraction_status"]
    ].head(10)
)
```

```python
#save extracted articles 
scraped_df.to_csv(
    "../data/scraped_articles.csv",
    index=False,
    encoding="utf-8-sig"
)

failed_df.to_csv(
    "../data/failed_articles.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Article datasets saved successfully!")
```
