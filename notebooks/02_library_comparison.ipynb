# Comparison of Web Scraping Libraries

## 1. BeautifulSoup

and XML documents. It is easy to learn and suitable for small
and medium-sized web scraping projects.

## 2. Scrapy

Scrapy is a Python web crawling framework designed for large-scale
data collection. It supports concurrent requests, automatic
request scheduling and structured data extraction.

# Web Scraping Performance Benchmark

## Objective

The objective of this experiment is to compare the HTML
parsing performance of BeautifulSoup and Scrapy.

Both libraries will process the same downloaded HTML pages.

We will measure:
1. Average processing time.
2. Total processing time.
3. Extracted text length.
4. Number of successful pages.

The experiment will be repeated 10 times to improve
the reliability of the timing results.

```python
import requests
import time
import statistics
import pandas as pd

from bs4 import BeautifulSoup
from scrapy import Selector

print("Benchmark libraries imported successfully!")
```

## Download sample webpages

```python
urls = [
    "https://en.wikipedia.org/wiki/Climate_change",
    "https://en.wikipedia.org/wiki/Renewable_energy",
    "https://en.wikipedia.org/wiki/Solar_energy",
    "https://en.wikipedia.org/wiki/Wind_power",
    "https://en.wikipedia.org/wiki/Greenhouse_gas"
]

headers = {
    "User-Agent": "StudentResearchProject/1.0 (educational use)"
}

pages = []

for url in urls:
    try:
        response = requests.get(
            url,
            headers=headers,
            timeout=20
        )

        response.raise_for_status()

        pages.append({
            "url": url,
            "html": response.text
        })

        print("Downloaded:", url)

    except requests.RequestException as e:
        print("Failed:", url, e)

print("\nSuccessfully downloaded:", len(pages))
```

## Benchmark BeautifulSoup

```python
def extract_beautifulsoup(html):

    soup = BeautifulSoup(html, "html.parser")

    text = soup.get_text(separator=" ", strip=True)

    return text
```

## Benchmark Scrapy

```python
def extract_scrapy(html):

    selector = Selector(text=html)

    text = " ".join(
        selector.xpath("//body//text()").getall()
    )

    return " ".join(text.split())
```

```python
#run benchmark 
results = []

repetitions = 10

for page in pages:

    html = page["html"]

    # BeautifulSoup benchmark
    bs_times = []

    for _ in range(repetitions):

        start = time.perf_counter()

        bs_text = extract_beautifulsoup(html)

        end = time.perf_counter()

        bs_times.append(end - start)

    # Scrapy benchmark
    scrapy_times = []

    for _ in range(repetitions):

        start = time.perf_counter()

        scrapy_text = extract_scrapy(html)

        end = time.perf_counter()

        scrapy_times.append(end - start)

    results.append({
        "URL": page["url"],
        "BeautifulSoup_Avg_Time": statistics.mean(bs_times),
        "Scrapy_Avg_Time": statistics.mean(scrapy_times),
        "BeautifulSoup_Text_Length": len(bs_text),
        "Scrapy_Text_Length": len(scrapy_text)
    })

benchmark_df = pd.DataFrame(results)

benchmark_df
```

## Calculate overall benchmark

```python
print("BEAUTIFULSOUP VS SCRAPY")

print(
    "\nBeautifulSoup average time:",
    benchmark_df["BeautifulSoup_Avg_Time"].mean()
)

print(
    "Scrapy average time:",
    benchmark_df["Scrapy_Avg_Time"].mean()
)

print(
    "\nBeautifulSoup total average:",
    benchmark_df["BeautifulSoup_Avg_Time"].sum()
)

print(
    "Scrapy total average:",
    benchmark_df["Scrapy_Avg_Time"].sum()
)
```

```python
#SAVE BENCHMARK RESULTS 
benchmark_df.to_csv(
    "../reports/benchmark_results.csv",
    index=False
)

print("Benchmark results saved successfully!")
```

## Benchmark Results and Discussion

The performance experiment was conducted using five publicly accessible Wikipedia pages related to climate change and renewable energy. Both BeautifulSoup and Scrapy processed the same downloaded HTML documents. Each extraction operation was repeated 10 times. 

The results demonstrated that BeautifulSoup had an average processing time of 0.2930 seconds per page, whereas Scrapy averaged 0.0810 seconds per page

Scrapy was approximately 3.62 times faster in this experiment. Both libraries successfully extracted text from all five pages 
The extracted text lengths differed because the two libraries handled HTML elements and whitespace differently.


Although Scrapy demonstrated better parsing performance in this experiment, BeautifulSoup was selected for the assignment because of its beginner-friendly interface, flexibility and suitability for collecting at least 50 articles.

This benchmark measures HTML parsing performance rather than complete crawling performance. Scrapy's concurrency and scheduling capabilities were not evaluated.
